```python
from tensorflow.keras.models import load_model

maskNet = load_model("mask_detector.h5", compile=False)

print("Mask model loaded successfully")
```

```python
faceNet = cv2.dnn.readNet(
    "deploy.prototxt",
    "res10_300x300_ssd_iter_140000.caffemodel"
)

print("Face detector loaded successfully")
```

```python
def detect_mask(image):

    frame = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)

    (h, w) = frame.shape[:2]

    blob = cv2.dnn.blobFromImage(
        frame,
        1.0,
        (300,300),
        (104.0,177.0,123.0)
    )

    faceNet.setInput(blob)
    detections = faceNet.forward()

    result = "No Face Found"

    for i in range(detections.shape[2]):

        confidence = detections[0,0,i,2]

        if confidence > 0.5:

            box = detections[0,0,i,3:7] * np.array([w,h,w,h])

            (startX,startY,endX,endY) = box.astype("int")

            startX=max(0,startX)
            startY=max(0,startY)
            endX=min(w-1,endX)
            endY=min(h-1,endY)

            face = frame[startY:endY,startX:endX]

            if face.size == 0:
                continue

            face = cv2.cvtColor(face,cv2.COLOR_BGR2RGB)
            face = cv2.resize(face,(224,224))

            face = img_to_array(face)
            face = preprocess_input(face)
            face = np.expand_dims(face,axis=0)

            (mask,withoutMask) = maskNet.predict(face,verbose=0)[0]

            if mask > withoutMask:
                result = "Mask Detected"
                color = (0,255,0)
            else:
                result = "No Mask Detected"
                color = (0,0,255)

            cv2.rectangle(
                frame,
                (startX,startY),
                (endX,endY),
                color,
                2
            )

            cv2.putText(
                frame,
                result,
                (startX,startY-10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                color,
                2
            )

    frame = cv2.cvtColor(frame,cv2.COLOR_BGR2RGB)

    return frame, result
```

```python
demo = gr.Interface(
    fn=detect_mask,
    inputs=gr.Image(type="numpy"),
    outputs=[
        gr.Image(),
        gr.Textbox()
    ],
    title="Face Mask Detection System",
    description="Upload image and detect mask"
)

demo.launch(share=True)
```

```python
import cv2
import numpy as np
import gradio as gr
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.preprocessing.image import img_to_array

def detect_mask(image):

    print("IMAGE TYPE:", type(image))

    if image is None:
        return None, "No image received"

    print("IMAGE SHAPE:", image.shape)

    frame = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)

    (h, w) = frame.shape[:2]

    blob = cv2.dnn.blobFromImage(
        frame,
        1.0,
        (300, 300),
        (104.0, 177.0, 123.0)
    )

    faceNet.setInput(blob)

    detections = faceNet.forward()

    print("Detections Shape:", detections.shape)

    result = "No Face Found"

    for i in range(detections.shape[2]):

        confidence = detections[0, 0, i, 2]

        if confidence > 0.5:

            box = detections[0, 0, i, 3:7] * np.array([w, h, w, h])

            (startX, startY, endX, endY) = box.astype("int")

            startX = max(0, startX)
            startY = max(0, startY)
            endX = min(w - 1, endX)
            endY = min(h - 1, endY)

            face = frame[startY:endY, startX:endX]

            if face.size == 0:
                continue

            face = cv2.cvtColor(face, cv2.COLOR_BGR2RGB)
            face = cv2.resize(face, (224, 224))

            face = img_to_array(face)
            face = preprocess_input(face)
            face = np.expand_dims(face, axis=0)

            try:
                (mask, withoutMask) = maskNet.predict(
                    face,
                    verbose=0
                )[0]

                if mask > withoutMask:
                    result = "Mask Detected - Access Allowed"
                    color = (0, 255, 0)
                else:
                    result = "No Mask Detected - Access Denied"
                    color = (0, 0, 255)

                cv2.rectangle(
                    frame,
                    (startX, startY),
                    (endX, endY),
                    color,
                    2
                )

                cv2.putText(
                    frame,
                    result,
                    (startX, startY - 10),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.7,
                    color,
                    2
                )

            except Exception as e:
                print("Prediction Error:", e)
                return image, str(e)

    frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    return frame, result


demo = gr.Interface(
    fn=detect_mask,
    inputs=gr.Image(
        type="numpy",
        sources=["webcam"]
    ),
    outputs=[
        gr.Image(label="Result"),
        gr.Textbox(label="Prediction")
    ],
    title="Face Mask Detection System",
    description="Real-Time Face Mask Detection using MobileNetV2"
)

demo.launch(share=True)
```
